### Collection of documents

```python
documents = {
    "D1": "Reading books can improve vocabulary, knowledge, creativity, and critical thinking skills.",
    
    "D2": "Machine learning allows computers to learn patterns from data and make useful predictions.",
    
    "D3": "Python is a popular programming language used for software development data analysis and automation.",
    
    "D4": "Regular exercise improves physical fitness, increases energy, and supports a healthy lifestyle.",
    
    "D5": "Information retrieval helps users find relevant information from a large collection of documents.",
    
    "D6": "Protecting the environment requires reducing pollution, saving water, and planting more trees.",
    
    "D7": "Deep learning uses neural networks to identify complex patterns in large datasets.",
    
    "D8": "Travel gives people opportunities to explore new places, experience different cultures, and learn new things."
}

print("Total documents:", len(documents))
```

### Text Preprocessing and tokenization

```python
import re

def tokenize(text):
    text = text.lower()
    return re.findall(r'\b[a-z]+\b', text)
```

```python
for doc_id, text in documents.items():
    print(doc_id, ":", tokenize(text))
```

### Creating the Dictonary 

```python
dictionary = set()

for text in documents.values():
    words = tokenize(text)
    dictionary.update(words)

dictionary = sorted(dictionary)

print("Dictionary:")
print(dictionary)
```

```python
print("Total unique words:", len(dictionary))
```

### Inverted Index

```python
inverted_index = {}

for doc_id, text in documents.items():
    words = tokenize(text)

    for word in words:
        if word not in inverted_index:
            inverted_index[word] = set()

        inverted_index[word].add(doc_id)

print("Inverted Index:")
```

```python
for word in sorted(inverted_index):
    print(word, ":", sorted(inverted_index[word]))
```

### Boolean Retrieval 

```python
# AND Operation 
def boolean_and(word1, word2):
    return inverted_index.get(word1, set()) & inverted_index.get(word2, set())
```

```python
print("machine AND learning:")
print(sorted(boolean_and("machine", "learning")))
```

```python
# OR Operation 
def boolean_or(word1, word2):
    return inverted_index.get(word1, set()) | inverted_index.get(word2, set())
```

```python
print("Query: python OR travel")
print("Result:", sorted(boolean_or("python", "travel")))
```

```python
# NOT Operation 
def boolean_not(word1, word2):
    return inverted_index.get(word1, set()) - inverted_index.get(word2, set())
```

```python
print("Query: learning NOT python")
print("Result:", sorted(boolean_not("learning", "python")))
```

### Boolean Retrievals Results 

```python
print("Query 1: machine AND learning")
print("Result:", sorted(boolean_and("machine", "learning")))

print()

print("Query 2: python OR travel")
print("Result:", sorted(boolean_or("python", "travel")))

print()

print("Query 3: learning NOT python")
print("Result:", sorted(boolean_not("learning", "python")))
```
